# RAG Pipeline Example

This notebook demonstrates a retrieval-augmented generation workflow using local documents, embeddings, and an LLM.

```python
import os
import pathlib
from pprint import pprint

from rag_pipeline import (
    TextChunk,
    load_documents,
    create_text_chunks,
    build_vector_index,
    answer_query,
)

BASE_DIR = pathlib.Path().resolve()
os.environ.setdefault('OPENAI_API_KEY', '')  # Set your API key in environment variables
print('Working directory:', BASE_DIR)
```

## 1. Load documents

Load any text, markdown, CSV, or JSON files from the current folder.

```python
documents = load_documents(BASE_DIR)
print(f'Loaded {len(documents)} documents')
for doc in documents:
    print('-', doc['path'])
```

## 2. Create text chunks

Split text into overlapping chunks for retrieval.

```python
chunks = create_text_chunks(documents, chunk_size=250, overlap=50)
print(f'Created {len(chunks)} chunks')
print('Sample chunk text:', chunks[0].text[:300] if chunks else 'No chunks')
```

## 3. Build vector index

Embed each chunk and prepare the retrieval index.

```python
build_vector_index(chunks, model='text-embedding-3-small')
print('Vector index built successfully')
```

## 4. Ask a question

Use the retriever and LLM to answer a question with document context.

```python
question = 'What is the grade for a student who scored 85?'
result = answer_query(question, chunks, retrieve_top_k=3)
pprint(result)
```

## 5. Common errors to watch for

- Missing or invalid `OPENAI_API_KEY`
- No local documents found
- Chunks not embedded before retrieval
- Model request failures or quota errors
